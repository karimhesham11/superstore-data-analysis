```python
df['Discount_Group'] = pd.cut(
    df['Discount'],
    bins=[-0.01, 0, 0.10, 0.20, 0.30, 0.40, 0.50],
    labels=['0%', '1-10%', '11-20%', '21-30%', '31-40%', '41-50%']
)
```

كات دي بتقسملك الحاجه لمجموعات فا قسمت الديسكاونت وحطيته ف فاريابول ديسكاونت جروب

```python
df.groupby('Discount_Group', observed=True)['Profit'].mean()
```

```python
summary['Loss_Rate'] = summary['Loss_Rate'] * 100
```

```python
summary
```

```python
Q1 = df['Sales'].quantile(0.25)
Q2 = df['Sales'].quantile(0.50)
Q3 = df['Sales'].quantile(0.75)

print("Q1:", Q1)
print("Median:", Q2)
print("Q3:", Q3)
```

النقط الي ظاهره ف البوكس بلوت دي outliers

```python
IQR = Q3 - Q1

print("IQR:", IQR)
```

```python
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
```

```python
monthly_sales_profit = df.groupby('Year_Month').agg(
    Total_Sales=('Sales', 'sum'),
    Total_Profit=('Profit', 'sum')
)

monthly_sales_profit.head(48)
```

ربطنا السيلز و البروفيت ب الشهور

```python
monthly_sales_profit['Profit_Margin'] = (
    monthly_sales_profit['Total_Profit']
    / monthly_sales_profit['Total_Sales']
)

monthly_sales_profit.head(10)
```

```python
monthly_sales_profit['Profit_Margin_%'] = (
    monthly_sales_profit['Profit_Margin'] * 100
)

monthly_sales_profit.head(48)
```

```python
monthly_sales_profit['Profit_Margin_%'].plot(kind='line')

plt.xlabel('Year-Month')
plt.ylabel('Profit Margin (%)')
plt.title('Monthly Profit Margin Trend')

plt.show()
```

Sales showed an overall upward trend, but monthly profitability fluctuated significantly. July 2014 and January 2015 recorded negative profit margins, highlighting that strong sales do not necessarily guarantee strong profitability.

-- -- -- insights -- -- -- -- 

The High Sales + High Profit segment is the strongest customer segment, generating approximately $890K in sales and $211K in profit across 133 customers, indicating strong profitability with relatively low customer concentration risk.

The High Sales + Low Profit segment generated approximately $181K in sales but resulted in a $35.4K loss, with a -19.6% profit margin. This indicates that high sales volume is not translating into profitability and may warrant further investigation into discounts, product mix, and order-level costs.

The Low Sales + Medium High Profit segment generates relatively low sales but maintains a strong 29.6% profit margin. This segment represents a potential growth opportunity by increasing customer spending while maintaining its current profitability.

The Medium High Sales + Low Profit segment generates $139K in sales but results in a $19.3K loss across 48 customers. Rather than focusing on increasing sales, the priority should be to identify and address the underlying drivers of negative profitability before pursuing further growth.

```python
import os

segments = customer_analysis[['Segment']].reset_index()
segments.columns = ['Customer_ID', 'Customer_Segment']

# تحقق قبل التصدير
print(segments.shape)                              # عدد الصفوف = عدد العملاء
print(segments['Customer_ID'].is_unique)           # لازم True
print(segments['Customer_Segment'].isna().sum())   # لازم 0
print(segments['Customer_Segment'].value_counts()) # لازم يطابق الأرقام عندك (133 و27 و48...)

segments.to_csv('customer_segments.csv', index=False, encoding='utf-8-sig')
print(os.path.abspath('customer_segments.csv'))    # مكان الملف
```

```python
def group(s):
    if s == 'High Sales + High Profit':
        return 'Stars'
    if s in ['High Sales + Low Profit', 'Medium High Sales + Low Profit']:
        return 'High Sales, Losing Money'
    if s in ['Low Sales + Medium High Profit', 'Medium Low Sales + High Profit']:
        return 'Small but Profitable'
    return 'Other'

segments['Segment_Group'] = segments['Customer_Segment'].apply(group)
print(segments['Segment_Group'].value_counts())

segments.to_csv('customer_segments.csv', index=False, encoding='utf-8-sig')
```
